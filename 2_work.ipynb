# Teaching Price Statistics

The calculations below use normalized item names and exclude missing prices, review-flagged rows, and label-collision rows.

```python
import pandas as pd

path = "teaching_prices_cleaned_fixed.csv"
data = pd.read_csv(path)
data["date"] = pd.to_datetime(data["date"])
data.head()
```

```python
valid = data.loc[
    ~data["price_missing_flag"]
    & ~data["price_review_flag"]
    & ~data["label_collision_flag"]
]

summary = valid.groupby("item_normalized_candidate")["price"].agg(
    mean="mean",
    median="median",
    count="count",
).round(6)
summary
```

```python
city_means = valid.groupby(["item_normalized_candidate", "date"])["price"].mean()
first = city_means.groupby(level=0).first()
last = city_means.groupby(level=0).last()
percentage_change = ((last / first - 1) * 100).round(6)
percentage_change
```

```python
for item in ["Staple A", "Sweetener D"]:
    item_data = data.loc[data["item_normalized_candidate"] == item]
    with_flagged = item_data["price"].mean()
    without_flagged = item_data.loc[
        ~item_data["price_missing_flag"] & ~item_data["price_review_flag"],
        "price"
    ].mean()
    print(item)
    print("mean with flagged:", round(with_flagged, 6))
    print("mean without flagged:", round(without_flagged, 6))
```
